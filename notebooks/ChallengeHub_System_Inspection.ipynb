# ChallengeHub — System Inspection

Use this notebook to connect to the course project stored in Google Drive and inspect the supplied prototype.

```python
# Connect Course Project
from pathlib import Path
import os
import sys

PROJECT_FOLDER = "ChallengeHub"

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    drive.mount("/content/drive")
    PROJECT_ROOT = Path("/content/drive/MyDrive") / PROJECT_FOLDER
else:
    # Local Jupyter fallback.
    current = Path.cwd().resolve()
    PROJECT_ROOT = next(
        (
            p for p in (current, *current.parents)
            if (p / "challengehub").is_dir()
            and (p / "data").is_dir()
            and (p / "notebooks").is_dir()
        ),
        None,
    )
    if PROJECT_ROOT is None:
        raise FileNotFoundError(
            f"{PROJECT_FOLDER} could not be located from the current local working directory."
        )

required = [
    PROJECT_ROOT / "challengehub",
    PROJECT_ROOT / "data",
    PROJECT_ROOT / "notebooks",
    PROJECT_ROOT / "outputs",
]
missing = [path for path in required if not path.exists()]

if missing:
    expected = f"My Drive/{PROJECT_FOLDER}" if IN_COLAB else str(PROJECT_ROOT)
    raise FileNotFoundError(
        "The ChallengeHub course project is not set up correctly.\n"
        f"Expected project root: {expected}\n"
        "Required folders: challengehub, data, notebooks, outputs."
    )

os.chdir(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_DIR = PROJECT_ROOT / "data"
NOTEBOOKS_DIR = PROJECT_ROOT / "notebooks"
OUTPUTS_DIR = PROJECT_ROOT / "outputs"

ASSESSMENT_OUTPUTS = {
    "M2LA2": OUTPUTS_DIR / "M2LA2",
    "PO4": OUTPUTS_DIR / "PO4",
    "M2CW": OUTPUTS_DIR / "M2CW",
    "Final_Project": OUTPUTS_DIR / "Final_Project",
}

for folder in ASSESSMENT_OUTPUTS.values():
    folder.mkdir(parents=True, exist_ok=True)

print("ChallengeHub connected.")
print("Project root:", PROJECT_ROOT)
print("Working directory:", Path.cwd())
```

```python
from challengehub import ChallengeManager

app = ChallengeManager(PROJECT_ROOT / "data")
print("Participants:", len(app.participants))
print("Challenges:", len(app.challenges))
print("Activities:", len(app.activities))
```
